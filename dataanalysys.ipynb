```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data={
    "Name":["Sambodhi","Gunjan","Akanksha","Roshani","Pranjal"],
    "Math":[90,80,70,60,88],
    "Science":[90,78,65,90,77],
    "English":[88,78,68,58,77],
}
df=pd.DataFrame(data)
print(df)
```

```python
marks=df[["Math","Science","English"]].values
average_marks=np.mean(marks,axis=1)
df["Average"]=average_marks
print(df)
```

```python
#let's assign the grade
def grade(avg):
    if avg>=90:
        return "A"
    elif avg>=75:
        return "B"
    elif avg>=60:
        return "C"
    else:
        return "D"

df['Grade'] =df["Average"].apply(grade)
```

```python
#let's plot the graph
plt.figure()
plt.bar(df["Name"], df["Average"])
plt.xlabel("Students")
plt.ylabel("Average Marks")
plt.title("Average Marks of Students")
plt.show()
```

```python
#average marksnper subject
df[["Math","Science","English"]].mean().plot(kind="bar")
plt.title("average_marks_per_subject")
plt.ylabel("Marks")
plt.show()
```
