```python
#import lib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#load IT data
df = pd.read_csv('iot_sensor_data.csv')
df['Timestamp'] = pd.to_datetime(df['Timestamp'])
print("IOT Sensor Data\n", df)
```

```python
# statistical analysis

temp = df['Temperature'].values
mean_temp = np.mean(temp)       # Mean temperature
std_temp = np.std(temp)         # Standard deviation
max_temp = np.max(temp)         # Maximum temperature
min_temp = np.min(temp)         # Minimum temperature
variance_temp = np.var(temp)    # Variance

print("\nMean Temperature: ", mean_temp)

print("\nStandard Deviation of Temperature: ", std_temp)

print("\nMaximum Temperature: ", max_temp)

print("\nMinimum Temperature: ", min_temp)

print("\nVariance of Temperature: ", variance_temp)
```

```python
#Plotting the temperature data
plt.figure()
plt.plot(df["Timestamp"], df["Temperature"], label='Temperature', color='orange')
plt.title('IoT Sensor Temperature Data')
plt.grid()
plt.xlabel('Timestamp')
plt.ylabel('Temperature (°C)')
plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.bar(
    range(len(df)),
    df["Temperature"],
    width=0.8,
    edgecolor="black"
)

plt.xlabel("Timestamp")
plt.ylabel("Temperature")
plt.title("Temperature Sensor Data")

# Show fewer timestamp labels
step = max(1, len(df) // 10)
plt.xticks(
    range(0, len(df), step),
    df["Timestamp"].dt.strftime("%H:%M:%S").iloc[::step],
    rotation=45
)

plt.tight_layout()
plt.show()
```
